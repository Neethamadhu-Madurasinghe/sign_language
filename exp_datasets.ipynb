```python
from pose_format import Pose
from pose_format.pose_visualizer import PoseVisualizer
from IPython.display import Image
from pose_format import PoseHeader
```

```python
pose_file = "all_outputs/output_pose/10967.pose"
output_folder = "all_outputs/output_pose"
```

```python
data_buffer = open(pose_file, "rb").read()
pose = Pose.read(data_buffer)
p = PoseHeader()

numpy_data = pose.body.data
confidence_measure  = pose.body.confidence
confidence_measure
```

```python
with open(pose_file, "rb") as f:
    pose = Pose.read(f.read())

v = PoseVisualizer(pose)

v.save_gif(output_folder + "/output.gif", v.draw())

display(Image(open(output_folder + "/output.gif",'rb').read()))
```

```python
with open(pose_file, "rb") as f:
    pose = Pose.read(f.read())




pose.normalize(pose.header.normalization_info(
    p1=("POSE_LANDMARKS", "RIGHT_SHOULDER"),
    p2=("POSE_LANDMARKS", "LEFT_SHOULDER")
))
v_aug = PoseVisualizer(pose)
v_aug.save_gif(output_folder + "/output_augemented.gif", v_aug.draw())

display(Image(open(output_folder + "/output_augemented.gif",'rb').read()))
```
