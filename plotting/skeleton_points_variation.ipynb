```python
import numpy as np
import matplotlib.pyplot as plt

# Data for the models
avg_classes = np.array([7.23, 12.16, 17.03])
model_61_accuracy = np.array([0.5385, 0.7385, 0.7693])
model_27_accuracy = np.array([0.5425, 0.7126, 0.7591])

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(avg_classes, model_61_accuracy, marker='o', label='61 Skeleton Points')
plt.plot(avg_classes, model_27_accuracy, marker='o', label='27 Skeleton Points')
plt.ylim(0, 1)

# Add labels and title
plt.xlabel('Average Number of Classes')
plt.ylabel('Accuracy')
plt.title('Accuracy difference between number of skeleton points')

# Add legend
plt.legend()

# Show the plot
plt.grid(True)
plt.show()
```
